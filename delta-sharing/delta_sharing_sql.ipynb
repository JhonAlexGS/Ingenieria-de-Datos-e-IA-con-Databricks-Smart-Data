# Delta Sharing - capa golden
Se comparten las tablas de KPIs de golden con un destinatario externo:

1. Se crea el share `share_instacart_golden` y se agregan las tablas.
2. Se crea el recipient `recipient_area_comercial`.
3. Se otorga SELECT sobre el share al recipient.

El recipient recibe un enlace de activacion para descargar sus credenciales y leer las tablas sin acceso al workspace.

```python
%sql
CREATE SHARE IF NOT EXISTS share_instacart_golden
  COMMENT 'KPIs de negocio Instacart (capa golden)';
```

```python
%sql
ALTER SHARE share_instacart_golden ADD TABLE catalog_prod.golden.kpi_departamentos WITH HISTORY;
ALTER SHARE share_instacart_golden ADD TABLE catalog_prod.golden.top_productos WITH HISTORY;
ALTER SHARE share_instacart_golden ADD TABLE catalog_prod.golden.actividad_dia_hora WITH HISTORY;
```

```python
%sql
CREATE RECIPIENT IF NOT EXISTS recipient_area_comercial
  COMMENT 'Área comercial - consumo externo de KPIs';
```

```python
%sql
GRANT SELECT ON SHARE share_instacart_golden TO RECIPIENT recipient_area_comercial;
```

```python
%sql
-- Validacion
SHOW ALL IN SHARE share_instacart_golden;
DESCRIBE RECIPIENT recipient_area_comercial;
```
