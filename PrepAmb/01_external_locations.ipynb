# Preparacion de ambiente - external locations
Se crea una external location por cada contenedor del storage (`raw`, `bronze`, `silver`, `golden` y `catalog`).
Todas usan la storage credential `cred_instacart_jhon_garcia_15_nov`, creada con el Access Connector (Managed Identity), por lo que no se usan claves del storage.

Antes de ejecutar se reemplaza `<storage_account>` por el nombre del storage.

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `exlt_raw`
  URL 'abfss://raw@<storage_account>.dfs.core.windows.net/'
  WITH (STORAGE CREDENTIAL `cred_instacart_jhon_garcia_15_nov`)
  COMMENT 'Capa RAW: archivos fuente CSV de Instacart';
```
